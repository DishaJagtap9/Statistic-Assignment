Q21. Coefficient of Variation

Two investment portfolios have the following statistics:

. Portfolio A: Mean return = 12%, Standard deviation = 2%
. Portfolio B: Mean return = 18%, Standard deviation = 4%

Calculate the coefficient of variation for each portfolio and determine which investment is more
consistent.




```python
# Portfolio A
mean_a = 12
std_a = 2

# Portfolio B
mean_b = 18
std_b = 4

# Coefficient of Variation (CV)
cv_a = (std_a / mean_a) * 100
cv_b = (std_b / mean_b) * 100

print(f"Portfolio A CV: {cv_a:.2f}%")
print(f"Portfolio B CV: {cv_b:.2f}%")

# Compare consistency
if cv_a < cv_b:
    print("Portfolio A is more consistent.")
elif cv_b < cv_a:
    print("Portfolio B is more consistent.")
else:
    print("Both portfolios are equally consistent.")
```

Q22. Effect of Adding a Constant

The observations are:

14, 18, 22, 26, 30

Without recalculating from scratch, determine the new:

1. Mean
2. Variance
3. Standard deviation

if 8 is added to every observation.

```python
import numpy as np

# Original data
data = np.array([14, 18, 22, 26, 30])

# Original statistics
mean = np.mean(data)
variance = np.var(data)      # Population variance
std = np.std(data)           # Population standard deviation

# Add 8 to every observation
new_data = data + 8

# New statistics
new_mean = np.mean(new_data)
new_variance = np.var(new_data)
new_std = np.std(new_data)

print("Original Mean:", mean)
print("Original Variance:", variance)
print("Original Standard Deviation:", std)

print("\nNew Mean:", new_mean)
print("New Variance:", new_variance)
print("New Standard Deviation:", new_std)
```

The following data represent the number of customers visiting a store.

Calculate:

1. Mean

```python
import numpy as np

# Values and frequencies
x = np.array([10, 15, 20, 25, 30])
f = np.array([3, 5, 8, 6, 3])

# Total frequency
N = np.sum(f)

# Mean
mean = np.sum(x * f) / N

# Population variance
variance = np.sum(f * (x - mean) ** 2) / N

# Population standard deviation
std = np.sqrt(variance)

print("Mean:", round(mean, 2))
print("Population Variance:", round(variance, 2))
print("Population Standard Deviation:", round(std, 2))
```

```python
import random

outcomes = [
    ("H", "H"),
    ("H", "T"),
    ("T", "H"),
    ("T", "T")
]

count_one_head = 0
count_two_heads = 0
count_atleast_one_head = 0

trials = 10000

for i in range(trials):
    result = random.choice(outcomes)

    if result.count("H") == 1:
        count_one_head += 1

    if result.count("H") == 2:
        count_two_heads += 1

    if result.count("H") >= 1:
        count_atleast_one_head += 1


print("Exactly one head:", count_one_head/trials)
print("Two heads:", count_two_heads/trials)
print("At least one head:", count_atleast_one_head/trials)
```

Q33. Three Coin Tosses

Three fair coins are tossed.

Find the probability of obtaining:

1. Exactly two Heads.
2. At most one Head.
3. No Heads.

```python
outcomes=["HHH", "HHT", "HTH", "THH",
    "HTT", "THT", "TTH", "TTT"]

total=len(outcomes)

exactly_two=0
at_most_one=0
no_heads=0

for outcomes in outcomes:
    heads=outcomes.count("H")

    if heads ==2:
        exactly_two +=1

    if heads<=1:
        at_most_one +=1

    if heads ==0:
        no_heads +=1

print("Exactly two head",exactly_two/total)
print("at most one head",at_most_one/total)
print("no heads",no_heads/total)
```

Q34. Rolling a Die

A fair die is rolled once.

Find the probability of obtaining:

1. An even number.
2. A number greater than 4.
3. A prime number.

```python
import random

trials = 10

even = 0
greater_than_4 = 0
prime = 0

for i in range(trials):
    roll = random.randint(1, 6)

    if roll % 2 == 0:
        even += 1

    if roll > 4:
        greater_than_4 += 1

    if roll in [2, 3, 5]:
        prime += 1

print("Even probability:", even / trials)
print("Greater than 4 probability:", greater_than_4 / trials)
print("Prime probability:", prime / trials)
```

```python
#Q96. A manufacturer claims that the average weight of a packet is 500 g. A sample of 64 packets has:
import math
from scipy.stats import norm

mu0 = 500          # claimed mean
n, xbar, sigma = 64, 495, 20
alpha = 0.05

# H0: mu = 500   H1: mu != 500  (two-tailed)
SE = sigma / math.sqrt(n)
Z = (xbar - mu0) / SE
z_crit = norm.ppf(1 - alpha/2)
p_value = 2 * (1 - norm.cdf(abs(Z)))

print("H0: mu = 500 g   |   H1: mu != 500 g")
print("Z statistic:", Z)
print("Critical Z value (two-tailed, alpha=0.05):", z_crit)
print("p-value:", p_value)

if abs(Z) > z_crit:
    print("Decision: Reject H0 (|Z| > z_critical)")
    print("Conclusion: There is sufficient evidence that the average packet weight differs from 500 g.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is not sufficient evidence that the average packet weight differs from 500 g.")
```

```python
#Q100. A machine is designed to fill bottles with 1 litre of liquid.
import math
from scipy.stats import norm

mu0 = 1.0
n, xbar, sigma = 100, 0.985, 0.04
alpha = 0.05

# H0: mu = 1 litre   H1: mu != 1 litre  (two-tailed)
SE = sigma / math.sqrt(n)
Z = (xbar - mu0) / SE
z_crit = norm.ppf(1 - alpha/2)
p_value = 2 * (1 - norm.cdf(abs(Z)))

print("H0: mu = 1.0 litre   |   H1: mu != 1.0 litre")
print("Z statistic:", Z)
print("Critical Z value (two-tailed, alpha=0.05):", z_crit)
print("p-value:", p_value)

if p_value < alpha:
    print("Decision: Reject H0 (p-value < alpha).")
else:
    print("Decision: Fail to reject H0 (p-value >= alpha).")

# 95% Confidence Interval
z_95 = norm.ppf(0.975)
margin = z_95 * SE
ci_lower, ci_upper = xbar - margin, xbar + margin
print(f"95% Confidence Interval for the population mean: ({ci_lower:.5f}, {ci_upper:.5f}) litres")

print()
print("Interpretation: We are 95% confident that the true mean fill volume lies within this interval.")
print("Since 1.0 litre is NOT contained in the interval, this confirms the machine is under-filling")
print("bottles on average, consistent with rejecting H0.")
```
